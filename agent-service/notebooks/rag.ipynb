# 5.向量化

```python
from app.rag.embedding import BGEM3Embedding

embedding = BGEM3Embedding()

query = "我爱上班"

vector = embedding.embed_query(query)
print(vector)
```

```python
print(embedding.dim()) # 稠密向量的维度
```

```python
# 向量化单条文本
text = "我爱上班"
vector = embedding.embed_query(text)['dense']

# 批量向量化
texts = ["我要躺平", "我爱工作", "拒绝加班"]
vectors = embedding.embed_documents(texts)['dense']

import numpy as np

def cosine_similarity(vec1, vec2):
    dot = np.dot(vec1, vec2)
    return dot / (np.linalg.norm(vec1) * np.linalg.norm(vec2))

for v in vectors:
    similarity = cosine_similarity(vector, v)
    print("Cosine Similarity:", similarity)
```

# 6.Milvus - Collection

```python
from pymilvus import MilvusClient, DataType

# 1.初始化客户端
client = MilvusClient(
    uri="http://192.168.150.101:19530"
)
```

```python

# 2.创建collection
dim = embedding.dim()

# 2.1.schema
schema = client.create_schema(
    auto_id=False, # 不自动生成id
    enable_dynamic_field=False, # 是否运行动态字段
)

schema.add_field(field_name="id", datatype=DataType.INT64, is_primary=True)
schema.add_field(field_name="content", datatype=DataType.VARCHAR, max_length=4096)
# 稠密向量字段
schema.add_field(field_name="dense_vector", datatype=DataType.FLOAT_VECTOR, dim=dim)
# 稀疏向量字段
schema.add_field(field_name="sparse_vector", datatype=DataType.SPARSE_FLOAT_VECTOR)

# 普通元数据字段，但是是数组类型
schema.add_field(
    field_name="section_path",
    datatype=DataType.ARRAY,
    element_type=DataType.VARCHAR,
    max_length=512,
    max_capacity=6,
)

# 2.2.索引
index_params = client.prepare_index_params()

index_params.add_index(
    field_name="id", # 要添加索引的字段名称
    index_type="AUTOINDEX"  # 索引的类型，AUTOINDEX就是由Milvus自己判断用什么索引
)

index_params.add_index(
    field_name="section_path",
    index_type="AUTOINDEX",
)
 # 稠密向量索引
index_params.add_index(
    field_name="dense_vector", # 字段名
    index_type="AUTOINDEX", # 稠密向量索引方式
    metric_type="COSINE" # 相似度计算方式，COSINE、IP
)
# 稀疏向量索引
index_params.add_index(
    field_name="sparse_vector",
    index_type="SPARSE_INVERTED_INDEX", # 稀疏向量，采用倒排索引
    metric_type="IP", # 相似度计算方式，支持BM25、IP
)

# 2.3.创建collection
client.create_collection(
    collection_name="my_collection_1",
    schema=schema,
    index_params=index_params
)

res = client.get_load_state(
    collection_name="my_collection_1"
)

print(res)


# 查看Collection
res = client.list_collections()

print(res)# ['my_collection_1']

```

# 8.Milvus-serach

```python
# 先把问题向量化
query = "等待期有多久？"
query_vector = embedding.embed_query(query)

print(query_vector['dense'])
```

## 8.1.稠密检索

```python
# 然后搜索：
res = client.search(
    collection_name="insurance_knowledge_chunks", # Collection名称
    anns_field="dense_vector", # 检索字段名（稠密）
    data=[query_vector['dense']], # 问题向量（稠密）
    limit=3, # 查询文档最大数量
    search_params={"metric_type": "COSINE"}, # 检索参数，比如向量相似度算法，必须和索引时一致
    output_fields=["content", "section_path"] # 要返回的字段名
)
for hits in res:
    for hit in hits:
        print(hit)
```

## 8.2.稀疏

```python
# 然后搜索：
res = client.search(
    collection_name="my_collection_1", # Collection名称
    anns_field="sparse_vector", # 检索字段名（稀疏）
    data=[query_vector['sparse']], # 问题向量（稀疏）
    limit=3, # 查询文档最大数量
    search_params={"metric_type": "IP"}, # 检索参数，比如向量相似度算法，必须和索引时一致
    output_fields=["content", "section_path"] # 要返回的字段名
)
for hits in res:
    for hit in hits:
        print(hit)
```

## 8.3.过滤

```python
# 然后搜索：
res = client.search(
    collection_name="my_collection_1", # Collection名称
    anns_field="sparse_vector", # 检索字段名（稀疏）
    data=[query_vector['sparse']], # 问题向量（稀疏）
    limit=3, # 查询文档最大数量
    search_params={"metric_type": "IP"}, # 检索参数，比如向量相似度算法，必须和索引时一致
    output_fields=["content", "section_path"], # 要返回的字段名
    filter="section_path[3] like '%等待期%'"
)
for hits in res:
    for hit in hits:
        print(hit)
```

## 8.3.混合检索

```python
from pymilvus import AnnSearchRequest, RRFRanker

# 1.创建稠密、稀疏请求
# 稠密
dense_request = AnnSearchRequest(
    data=[query_vector['dense']],
    anns_field="dense_vector",
    param={"metric_type": "COSINE"},
    filter="ARRAY_CONTAINS(product_ids, 1)",
    limit=3,
)

# 稀疏
sparse_request = AnnSearchRequest(
    data=[query_vector['sparse']],
    anns_field="sparse_vector",
    param={"metric_type": "IP"},
    filter="ARRAY_CONTAINS(product_ids, 1)",
    limit=3,
)

reqs = [dense_request, sparse_request]

# 2.准备混合策略
ranker = RRFRanker(k=60)

# 3.请求
res = client.hybrid_search(
    collection_name=settings.milvus_collection,
    reqs=reqs,
    ranker=ranker,
    limit=3,
    output_fields=['content', 'section_path']
)
for hits in res:
    print("TopK results:")
    for hit in hits:
        print(hit)
```
